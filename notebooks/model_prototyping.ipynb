# Model Prototyping for ICU Outcome Prediction
This notebook loads the ICU dataset, preprocesses it, trains a Random Forest classifier, evaluates the model, and visualizes feature importance using SHAP.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score
import shap
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data
df = pd.read_csv("../data/icu_extended_data.csv")
df.head()
```

```python
# Encode categorical data
df["gender"] = df["gender"].map({"Male": 0, "Female": 1})
```

```python
# Split features and target
X = df.drop("outcome", axis=1)
y = df["outcome"]

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
```

```python
# Evaluate model
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
```

```python
# SHAP explainability
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_test)

shap.summary_plot(shap_values[1], X_test, plot_type="bar")
```
